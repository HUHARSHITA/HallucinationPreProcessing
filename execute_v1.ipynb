```python
import json
from collections import Counter

# Convert answer column from JSON strings to dictionaries
trivia_answers = trivia_df["answer"].apply(json.loads)

# Count answer categories
answer_types = Counter(
    ans.get("type") for ans in trivia_answers
)

print("TriviaQA Answer Type Distribution:\n")

for category, count in answer_types.most_common():
    print(category, ":", count)
```

```python
alias_counts = [
    len(ans.get("aliases", []))
    for ans in trivia_answers
]

print("Average aliases:", sum(alias_counts) / len(alias_counts))
print("Maximum aliases:", max(alias_counts))
print("Minimum aliases:", min(alias_counts))
```

```python
nq_answers = nq_df["answer"].apply(json.loads)

nq_answer_counts = nq_answers.apply(len)

print("Average acceptable answers:", nq_answer_counts.mean())
print("Maximum answers:", nq_answer_counts.max())
print("Minimum answers:", nq_answer_counts.min())
```

```python
# Find the record with the maximum number of aliases

max_alias_idx = max(
    range(len(trivia_answers)),
    key=lambda i: len(trivia_answers.iloc[i].get("aliases", []))
)

record = trivia_answers.iloc[max_alias_idx]

print("Question:")
print(trivia_df.iloc[max_alias_idx]["question"])

print("\nPrimary Answer:")
print(record.get("value"))

print("\nAnswer Type:")
print(record.get("type"))

print("\nNumber of Aliases:")
print(len(record.get("aliases", [])))

print("\nFirst 20 Aliases:")
print(record.get("aliases", [])[:20])
```

```python
# Calculate alias count for every TriviaQA record

alias_counts = trivia_answers.apply(
    lambda x: len(x.get("aliases", []))
)

print("Total records:", len(alias_counts))
print("Records with more than 20 aliases:", (alias_counts > 20).sum())
print("Records with more than 50 aliases:", (alias_counts > 50).sum())
print("Records with more than 100 aliases:", (alias_counts > 100).sum())

print("\nAlias count statistics:")
print(alias_counts.describe(percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]))
```

```python
# Question length analysis
trivia_df["question_length"] = (
    trivia_df["question"].astype(str).str.split().str.len()
)

nq_df["question_length"] = (
    nq_df["question"].astype(str).str.split().str.len()
)

print("TRIVIAQA QUESTION LENGTH")
print(trivia_df["question_length"].describe())

print("\nNQ-OPEN QUESTION LENGTH")
print(nq_df["question_length"].describe())
```

```python
import json
import pandas as pd

# Preserve original datasets
trivia_processed = trivia_df.copy()
nq_processed = nq_df.copy()

# Parse TriviaQA answers
trivia_parsed = trivia_processed["answer"].apply(json.loads)

trivia_processed["reference_answer"] = trivia_parsed.apply(
    lambda x: x["value"]
)

trivia_processed["acceptable_answers"] = trivia_parsed.apply(
    lambda x: x["aliases"]
)

trivia_processed["answer_type"] = trivia_parsed.apply(
    lambda x: x["type"]
)

# Parse NQ-Open answers
nq_processed["acceptable_answers"] = nq_processed["answer"].apply(
    json.loads
)

nq_processed["reference_answer"] = nq_processed[
    "acceptable_answers"
].apply(lambda x: x[0] if x else None)

# Save processed datasets
trivia_processed.to_csv("TriviaQA_25K_Processed.csv", index=False)
nq_processed.to_csv("NQ_Open_25K_Processed.csv", index=False)

print("Processed datasets saved successfully.")
```
