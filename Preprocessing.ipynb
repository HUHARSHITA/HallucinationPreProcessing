```python
import os

# Get current working directory
save_dir = os.getcwd()

print("Current working directory:", save_dir)
```

```python
from datasets import load_dataset

trivia = load_dataset(
    "mandarjoshi/trivia_qa",
    "rc.nocontext",
    split="train",
    streaming=True
)

print("Dataset loaded successfully!")
```

```python
trivia = trivia.shuffle(seed=42, buffer_size=10000)

trivia_data = list(trivia.take(25000))

print("Number of records:", len(trivia_data))
print("Available attributes:", list(trivia_data[0].keys()))
```

```python
import pandas as pd
import json

trivia_df = pd.DataFrame(trivia_data)

# Convert nested lists/dictionaries into JSON strings
for col in trivia_df.columns:
    trivia_df[col] = trivia_df[col].apply(
        lambda x: json.dumps(x, ensure_ascii=False)
        if isinstance(x, (dict, list))
        else x
    )

trivia_df.to_csv(
    os.path.join(save_dir, "TriviaQA_25K.csv"),
    index=False,
    encoding="utf-8-sig"
)

print("TriviaQA saved successfully!")
print("Shape:", trivia_df.shape)
print("Columns:", list(trivia_df.columns))
```

```python
from datasets import load_dataset

nq = load_dataset(
    "google-research-datasets/nq_open",
    split="train",
    streaming=True
)

print("Natural Questions loaded successfully!")
```

```python
nq = nq.shuffle(seed=42, buffer_size=10000)
nq_data = list(nq.take(25000))

print("Number of records:", len(nq_data))
print("Available attributes:", list(nq_data[0].keys()))
```
